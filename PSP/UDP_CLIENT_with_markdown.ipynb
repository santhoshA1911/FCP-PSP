# 🖥️ UDP Client

### 📌 Objective
Create a UDP client using Python's `socket` module and communicate with a UDP server through **127.0.0.1:5000**.

## ⚙️ How It Works

🔹 **1. Create Socket** → Uses IPv4 and UDP.

🔹 **2. Set Server Address** → Stores the server IP address and port number.

🔹 **3. Send Message** → Sends `"Hello Server"` using `sendto()`.

🔹 **4. Receive Reply** → Gets the server response using `recvfrom()`.

🔹 **5. Decode** → Converts received bytes into readable text.

🔹 **6. Close** → Closes the client socket.

## 🔄 Communication

<div style="background:#ecfff7; border-radius:12px; padding:16px;">

👤 **UDP Client**  →  🖥️ **UDP Server**

🖥️ **UDP Server**  →  💬 **Response to Client**

</div>

💡 **Key Point:** UDP is connectionless, so the client sends data directly to the server without establishing a permanent connection.

</div>


```python
import socket

# Create UDP socket
client = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)

# Server IP address and port
server_address = ("127.0.0.1", 5000)

# Send message to server
client.sendto("Hello Server".encode(), server_address)

# Receive response from server
data, address = client.recvfrom(1024)

# Display server response
print("Server:", data.decode())

# Close client socket
client.close()
```
