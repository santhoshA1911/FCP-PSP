<h1 style="color:#3730a3;">🖥️ UDP SERVER</h1>
<p style="font-size:16px;"><b>📡 Purpose:</b> Receives a message from the UDP Client and sends a response back.</p>

<h2 style="color:#7c3aed;">⚙️ How It Works</h2>
<p>🔹 <b>1. Create Socket</b> → Uses IPv4 and UDP.</p>
<p>🔹 <b>2. Bind</b> → Connects the server to <code>127.0.0.1:5000</code>.</p>
<p>🔹 <b>3. Receive</b> → Gets data using <code>recvfrom()</code>.</p>
<p>🔹 <b>4. Decode</b> → Converts bytes into readable text.</p>
<p>🔹 <b>5. Reply</b> → Sends a response using <code>sendto()</code>.</p>
<p>🔹 <b>6. Close</b> → Closes the server socket.</p>

<h2 style="color:#059669;">🔄 Communication</h2>
<div style="padding:12px; border-radius:10px; background:#ecfdf5; font-size:17px;">
👤 <b>Client</b> &nbsp; ➜ &nbsp; 📡 <b>UDP Server</b><br><br>
📡 <b>UDP Server</b> &nbsp; ➜ &nbsp; 💬 <b>Response to Client</b>
</div>

<p style="margin-top:15px;"><b>💡 Key Point:</b> UDP is connectionless, so the server does not establish a permanent connection with the client.</p>
</div>

```python
import socket

server = socket.socket(socket.AF_INET, socket.SOCK_DGRAM)

server.bind(("127.0.0.1", 5000))

print("UDP Server is running...")

data, address = server.recvfrom(1024)

print("Client:", data.decode())

server.sendto("Hello Client".encode(), address)

server.close()
```
