```python
import sqlite3

conn = sqlite3.connect("college.db")
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE Department (
    Dept_ID INTEGER PRIMARY KEY,
    Department_Name TEXT
)
""")

cursor.execute("""
CREATE TABLE Student (
    Student_ID INTEGER PRIMARY KEY,
    Name TEXT,
    Age INTEGER,
    Dept_ID INTEGER,
    FOREIGN KEY (Dept_ID) REFERENCES Department(Dept_ID)
)
""")

print("Tables created successfully!")
```

```python
cursor.execute("INSERT INTO Department VALUES (1, 'IT')")
cursor.execute("INSERT INTO Department VALUES (2, 'CSE')")

cursor.execute("INSERT INTO Student VALUES (101, 'Sam', 20, 1)")
cursor.execute("INSERT INTO Student VALUES (102, 'Ram', 21, 2)")
cursor.execute("INSERT INTO Student VALUES (103, 'John', 20, 1)")

conn.commit()

print("Data inserted successfully!")
```

```python
cursor.execute("SELECT * FROM Student")

for row in cursor.fetchall():
    print(row)
```

```python
cursor.execute("""
SELECT Student.Name, Student.Age, Department.Department_Name
FROM Student
JOIN Department
ON Student.Dept_ID = Department.Dept_ID
""")

for row in cursor.fetchall():
    print(row)
```
